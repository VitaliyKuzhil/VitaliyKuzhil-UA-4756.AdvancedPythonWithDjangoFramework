# Завдання 1
**Суть завдання:** Для заданого числа `n`, потрібно знайти `k`-те число у зростаючій послідовності, що складається з усіх додатних чисел, які є або степенями `n`, або сумами різних степенів `n`.

**Пояснення:**
- Степінь числа `n` - це числа виду `n^0, n^1, n^2, ...`.
- Суми різних степенів `n` - це числа, які можна отримати, додаючи різні степені `n`.
- Наприклад, для `n = 3`, послідовність починається як `1, 3, 4, 9, 10, 12, 13...`.

**Принцип розв'язку:**
1. Генеруйте всі можливі числа, які є степенями або сумами різних степенів `n`.
2. Використовуйте структуру даних, як-от множину або пріоритетну чергу, щоб зберігати ці числа у відсортованому порядку.
3. Знайдіть `k`-те число у відсортованій послідовності.

# Завдання 6
**Суть завдання:** Визначити, чи є масив цілих чисел відсортованим у "зростаючому" порядку, "спадному" порядку або "не відсортованим" взагалі.

**Пояснення:**
- Зростаючий порядок - кожен наступний елемент більше або дорівнює попередньому.
- Спадний порядок - кожен наступний елемент менше або дорівнює попередньому.
- Не відсортований - якщо не відповідає жодному з попередніх двох.

**Принцип розв'язку:**
1. Перевірте, чи кожен елемент масиву відповідає одному з трьох можливих варіантів.

# Завдання 7
**Суть завдання:** Обчислити кількість балів, які Ніккі отримає як подарунок, враховуючи видимі нулі у числі `N`, і повернути це число у двійковій формі.

**Пояснення:**
- Видимі нулі обчислюються так: 0, 6 і 9 містять по 1 видимому нулю кожен; 8 містить 2 видимих нулі.
- Якщо кількість видимих нулів (M) парна і більше 0, Ніккі повинен віддати один бал компанії. Якщо M непарна, компанія дає Ніккі один додатковий бал.

**Принцип розв'язку:**
1. Порахуйте видимі нулі в числі `N`.
2. Визначте кількість балів, які отримає Ніккі.
3. Поверніть це число у двійковій формі.
